# Linear Algebra
===============

Linear Algebra is a Branch Of Mathematics concerned with the algebric properties of mathematical systems that genaralize the properties of vector space.

From Engineering perspective, we view linear spaces as signal spaces and linear transformations that are used as a mathamtical modles for systems.

## Linear Spaces

A vector can also be written as an equation, for example:

\[
    x(t) = \sin(t) + \cos(t)
    \]

    To write equations in Markdown with LaTeX:

    - Inline equation: `$x(t) = \sin(t) + \cos(t)$`
    - Display equation:

        $$
        x(t) = \sin(t) + \cos(t)
        $$
